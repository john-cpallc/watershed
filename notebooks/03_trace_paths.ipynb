# Trace raindrop paths

Use the same tracer as the web API.

```python
import geopandas as gpd
import matplotlib.pyplot as plt
from watershed.geo import PROCESSED_DIR, WGS84, load_county_boundary
from watershed.trace import FlowTracer

tracer = FlowTracer(PROCESSED_DIR)
county = load_county_boundary()
points = {'north': (-88.20, 31.00), 'inland': (-88.15, 30.85), 'south': (-88.10, 30.26)}
results = {name: tracer.trace(*xy) for name, xy in points.items()}
for name, r in results.items():
    print(name, r.terminated_at, round(r.length_km, 1), 'km', r.cells_traversed)
```

```python
fig, ax = plt.subplots(figsize=(10, 10))
county.boundary.plot(ax=ax, color='#0369a1', linewidth=1.5)
for name, r in results.items():
    gpd.GeoSeries([r.line_wgs84], crs=WGS84).plot(ax=ax, linewidth=2, label=name)
ax.legend()
ax.set_title('Traced paths')
ax.set_axis_off()
plt.tight_layout()
```
