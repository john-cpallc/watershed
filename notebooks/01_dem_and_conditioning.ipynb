# DEM conditioning and smoothing

Load the Jefferson County DEM, compare raw vs smoothed surfaces, and run depression filling.

```python
from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np
from watershed.hydro import fill_depressions, read_geotiff, resolve_flats

ROOT = Path('..').resolve()
raw, _, _ = read_geotiff(ROOT / 'data/processed/dem_10m_raw.tif')
smooth, _, _ = read_geotiff(ROOT / 'data/processed/dem_10m_smooth.tif')
filled = fill_depressions(smooth)
conditioned = resolve_flats(filled)
print('raised cells', int(np.sum(filled > smooth + 1e-6)))
```

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4))
for ax, data, title in zip(axes, [raw, smooth, smooth - raw], ['Raw', 'Smoothed', 'Diff']):
    im = ax.imshow(data, cmap='terrain')
    ax.set_title(title)
    ax.axis('off')
    fig.colorbar(im, ax=ax, fraction=0.046)
plt.tight_layout()
```
