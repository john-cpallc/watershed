# D8 flow direction and accumulation

Visualize the discrete vector field (D8) and where channels form.

```python
from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np
import rasterio

ROOT = Path('..').resolve()
with rasterio.open(ROOT / 'data/processed/fdir.tif') as src:
    fdir = src.read(1)
with rasterio.open(ROOT / 'data/processed/acc.tif') as src:
    acc = src.read(1)
print('max accumulation', int(acc.max()))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].imshow(fdir, cmap='tab20')
axes[0].set_title('Flow direction')
axes[1].imshow(np.log10(acc + 1), cmap='Blues')
axes[1].set_title('log10(accumulation)')
for ax in axes:
    ax.axis('off')
plt.tight_layout()
```
